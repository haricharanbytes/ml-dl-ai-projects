### Reverse of an Array

```python
def reverse(arr):

    left = 0
    right = len(arr) - 1 

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left = left + 1
        right = right - 1
    return arr


arr = [9, 8, 7, 6, 5, 4, 3, 2, 1]
res = reverse(arr)
print(res) 
```

### Duplicates in Array

```python
def findDuplicate(arr):

    org = set()
    dup = set()

    for item in arr:
        if item in org:
            dup.add(item)
        else:
            org.add(item)
    return dup

def main():
    arr = [1,2,3,4,5,2,4,1]
    dup = findDuplicate(arr)
    result = list(dup)
    print(result)

if __name__ == '__main__':
    main()
```
