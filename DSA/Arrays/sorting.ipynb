## Bubble sort

```python
def bubblesort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break

arr = [44,64,23,677,1,332]
bubblesort(arr)
print(arr)
```

## Selection sort

```python
def selection(arr):
    for i in range(len(arr)):
        min = i

        for j in range(i+1, len(arr)):
            if arr[j] < arr[min]:
                min = j
        arr[i], arr[min] = arr[min], arr[i]

arr = [1,4,2,6,3]
selection(arr)
arr
```
